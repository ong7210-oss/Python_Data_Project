```python
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns

#loading data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#data cleanup
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA_US = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'United States')]
```

```python
df_DA_US['job_posted_month_no'] = df_DA_US['job_posted_date'].dt.month
```

```python
df_DA_US_explode = df_DA_US.explode('job_skills')
```

```python
df_DA_US_pivot = df_DA_US_explode.pivot_table(index='job_posted_month_no', columns='job_skills', aggfunc='size', fill_value=0)

df_DA_US_pivot.loc['Total'] = df_DA_US_pivot.sum()

df_DA_US_pivot = df_DA_US_pivot[df_DA_US_pivot.loc['Total'].sort_values(ascending=False).index]

df_DA_US_pivot = df_DA_US_pivot.drop('Total')

df_DA_US_pivot
```

```python
DA_total = df_DA_US.groupby('job_posted_month_no').size()

DA_total
```

```python
df_DA_US_percent = df_DA_US_pivot.div(DA_total/100, axis=0)

df_DA_US_percent
```

```python
df_DA_US_percent = df_DA_US_percent.reset_index()

df_DA_US_percent['job_posted_month'] = df_DA_US_percent['job_posted_month_no'].apply(lambda x:pd.to_datetime(x, format='%m').strftime('%b'))

df_DA_US_percent = df_DA_US_percent.set_index('job_posted_month')

df_DA_US_percent = df_DA_US_percent.drop(columns = 'job_posted_month_no')

df_DA_US_percent
```

```python
df_plot_5 = df_DA_US_percent.iloc[:, :5]
```

```python
sns.lineplot(data=df_plot_5, dashes=False, palette='tab10')
sns.set_theme(style='ticks')
sns.despine()

plt.title('Trending top skills for Data Analyst in US')
plt.xlabel('2023')
plt.ylabel('Likehood in Job Posting')
plt.legend('')

from matplotlib.ticker import PercentFormatter
ax = plt.gca()
ax.yaxis.set_major_formatter(PercentFormatter(decimals=0))


for i in range(5):
    plt.text(11.5, df_plot_5.iloc[-1, i], df_plot_5.columns[i])   #x,y, text

plt.show()
```
